## Problema: intercambio catiónico en un acuífero detrítico

Se evalúa si la cantidad de intercambiador catiónico X calculada a partir de las propiedades del acuífero corresponde a 4 mmoles.

Datos:

- Densidad de conjunto: ρb = 1.86 g/cm³
- Porosidad: θ = 0.35
- Capacidad de intercambio catiónico: CEC = 75 meq/100 g

```python
rho_b = 1.86      # g/cm3
theta = 0.35      # porosidad
CEC = 75          # meq/100 g

# Convertir CEC a meq/g
CEC_meq_g = CEC / 100

# Masa de sólido por litro de agua poral
masa_solido_g = (rho_b / theta) * 1000

# Capacidad total de intercambio
X_meq = masa_solido_g * CEC_meq_g

# Convertir meq a equivalentes
X_eq = X_meq / 1000

# Convertir a mmol de carga equivalente
X_mmol = X_eq * 1000

print("Masa de sólido por L de agua poral:", round(masa_solido_g, 2), "g")
print("Intercambiador X:", round(X_meq, 2), "meq")
print("Intercambiador X:", round(X_eq, 3), "eq")
print("Intercambiador X:", round(X_mmol, 2), "mmol equivalentes")
```

## Resultado

La masa de sólido asociada a 1 L de agua poral es aproximadamente 5314.29 g.

Con una CEC de 0.75 meq/g, la capacidad total de intercambio es:

\[
X = 5314.29 \times 0.75 = 3985.71 \, \text{meq}
\]

Esto equivale a:

\[
X = 3.986 \, \text{eq}
\]

o aproximadamente:

\[
X = 3985.71 \, \text{mmol equivalentes}
\]

## Interpretación

La afirmación de que los moles de intercambiador X son 4 mmoles es falsa.

El cálculo muestra que la cantidad de intercambiador es aproximadamente 4 equivalentes, es decir, cerca de 4000 mmol equivalentes.

Por tanto, 4 mmoles subestima fuertemente la capacidad de intercambio catiónico del acuífero.

La afirmación es falsa. El valor calculado es cercano a 4 equivalentes, no a 4 mmoles.

Esto equivale aproximadamente a 4000 mmol equivalentes.

Por tanto, el acuífero tiene una capacidad de intercambio mucho mayor que la planteada en la afirmación.
