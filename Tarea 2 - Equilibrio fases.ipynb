## Resultados: Sistema abierto vs cerrado

| Sistema | pH final | Ca²⁺ (mol/kg) | HCO₃⁻ (mol/kg) |
|--------|---------|---------------|----------------|
| Cerrado | 9.910 | 1.069e-04 | 8.316e-05 |
| Abierto | 8.279 | 4.829e-04 | 9.496e-04 |

El sistema abierto disuelve más calcita. Esto se evidencia en mayores concentraciones de Ca2+ y HCO3-.

La presencia de CO2 reduce el pH (8.279 vs 9.910), favoreciendo la disolución del mineral.

En el sistema cerrado, el pH es más alto y la disolución de calcita es menor.

Por tanto, el equilibrio con CO2 atmosférico incrementa la solubilidad de la calcita.
