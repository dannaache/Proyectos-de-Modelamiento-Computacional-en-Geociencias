## Escenarios simulados

- Escenario actual: log PCO₂ = -3.5 (~420 ppm)

| Escenario | CO₂ atm aprox. | log P(CO₂) | pH | CO₂(aq) (mol/kg) | HCO₃⁻ (mol/kg) | CO₃²⁻ (mol/kg) | SI Aragonita | SI Calcita |

El aumento de CO2 atmosférico provoca una disminución progresiva del pH (de 8.357 a 7.928), evidenciando un proceso de acidificación oceánica.

El CO2 disuelto y el HCO3- aumentan con el incremento de CO2, mientras que el CO3-2 disminuye, indicando un desplazamiento del equilibrio del sistema carbonato hacia especies más protonadas.

Los índices de saturación de aragonita y calcita disminuyen, lo que sugiere menor estabilidad de minerales carbonatados.

Afecta la química del agua de mar y puede impactar organismos y sistemas carbonatados.
