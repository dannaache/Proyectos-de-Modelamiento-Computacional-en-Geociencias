```python
# ============================================
# Tarea de Modelamiento Computacional
# ============================================

# EJERCICIO 1 – Crear listas a partir de datos
# La idea es pasar la tabla a dos listas: meses (str) y precipitación (float)

precip_months = ["jan", "feb", "mar", "apr", "may", "jun", "jul", "aug", "sept", "oct", "nov", "dec"]

precip_inches = [0.70, 0.75, 1.85, 2.93, 3.05, 2.02, 1.93, 1.62, 1.84, 1.31, 1.39, 0.84]

# Chequeo rápido (para no embarrarla con el orden)
print("Meses:", precip_months)
print("Pulgadas:", precip_inches)
print("Cantidad de meses:", len(precip_months))
print("Cantidad de valores:", len(precip_inches))
```

```python
# EJERCICIO 2 – Conversión de unidades en una lista
# Vamos a convertir pulgadas a milímetros SIN dañar la lista original.

inch_to_mm = 25.4  # 1 pulgada = 25.4 mm

# Copia de la lista (importantísimo: no queremos cambiar precip_inches)
precip_mm = precip_inches.copy()

# Convertimos cada valor en la copia
for i in range(len(precip_mm)):
    precip_mm[i] = precip_mm[i] * inch_to_mm

print("Original (pulgadas):", precip_inches)
print("Convertida (mm):", precip_mm)
```

```python
# EJERCICIO 3 – Crear una lista de listas
# Agrupamos meses y precipitación (mm) en una sola estructura.

all_precip_data = [precip_months, precip_mm]

# Verificación
print(all_precip_data)
print("\nSublista 0 (meses):", all_precip_data[0])
print("Sublista 1 (mm):", all_precip_data[1])
```

```python
# 1) ¿Qué ventajas tiene agrupar datos relacionados en una sola estructura?
# Mantienes todo junto y organizado: meses con su precipitación. 
# Es más fácil pasar esto después a NumPy o pandas sin reescribir todo.

# 2) ¿Qué limitaciones tiene una lista de listas frente a una tabla?
# - No tiene nombres de columnas (toca acordarse qué es [0] y qué es [1]).
# - No es tan cómodo filtrar
```

```python
# EJERCICIO 4 – Graficar datos desde una lista de listas
# Gráfica de barras con matplotlib usando all_precip_data.

import matplotlib.pyplot as plt

# Crear figura y ejes (la agrandé un poquito para que se lea bien)
fig, ax = plt.subplots(figsize=(10, 5))

# Datos sacados desde la lista de listas
months = all_precip_data[0]
precip_values_mm = all_precip_data[1]

# Graficar
ax.bar(
    months,
    precip_values_mm,
    color="aqua"   # tono azul (puedes cambiarlo a "steelblue" si quieres algo menos brillante)
)

# Configurar títulos y etiquetas
ax.set(
    title="Precipitación media mensual (convertida a mm)",
    xlabel="Mes",
    ylabel="Precipitación (mm)"
)

# Extra para que no se monten las etiquetas si se ven apretadas
plt.xticks(rotation=45)

plt.show()
```
