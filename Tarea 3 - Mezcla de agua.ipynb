La mezcla genera condiciones tanto de subsaturación como sobresaturación. Para valores de SI < 0 (ej. 30–70%), ocurre disolución de calcita, lo que indica procesos de karstificación.

Para valores de SI > 0 (ej. >80%), se favorece la precipitación. Esto demuestra que la mezcla de aguas puede generar desequilibrio incluso si las aguas originales están cercanas al equilibrio.

La mezcla puede generar subsaturación y favorecer karstificación. Aunque las aguas originales no necesariamente están subsaturadas, la mezcla rompe el equilibrio químico. La zona de mezcla es donde ocurre la disolución

La mezcla muestra subsaturación en calcita principalmente entre 0% y 50% de agua de mar. En ese intervalo, el agua tendería a disolver calcita,
favoreciendo procesos de karstificación litoral.

A partir de ~60% de agua de mar, el SI de calcita es positivo, por lo que el sistema tiende a sobresaturarse.

El acuífero está apenas subsaturado, mientras que el agua de mar está sobresaturada en calcita.
